```python
# Paste this entire file into a single Kaggle notebook cell and run.
# It is self-contained and writes deliverables under /kaggle/working/practical6_outputs

import os, time, random, math, warnings
warnings.filterwarnings("ignore")
os.environ["CUDA_LAUNCH_BLOCKING"] = "1"  # helpful for debugging device-side assert

def _ensure(pkg, imp=None):
    try:
        __import__(imp or pkg)
    except Exception:
        import subprocess, sys
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])

for p in [("umap-learn","umap"), ("torchmetrics","torchmetrics"), ("scikit-learn","sklearn")]:
    _ensure(p[0], p[1])

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset, Subset, TensorDataset
from torchvision import datasets, transforms, utils as vutils
from torchvision.models import inception_v3
from sklearn.manifold import TSNE
import umap
from torchmetrics.functional.image import structural_similarity_index_measure

# =========================
# Config (edit only here)
# =========================
IS_KAGGLE = os.path.exists("/kaggle/working")
DATA_ROOT = "/kaggle/input" if IS_KAGGLE else "./data"
DOWNLOAD_CACHE = "/kaggle/working/data" if IS_KAGGLE else "./data"
DOWNLOAD_IF_MISSING = True
OUT_DIR = "/kaggle/working/lab6_outputs" if IS_KAGGLE else "./lav6_outputs"
os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(DOWNLOAD_CACHE, exist_ok=True)
FIG_DIR = os.path.join(OUT_DIR, "figures")
TAB_DIR = os.path.join(OUT_DIR, "tables")
for _d in [FIG_DIR, TAB_DIR]:
    os.makedirs(_d, exist_ok=True)

CFG = {
    "seed": 42,
    "batch_size": 128,
    "lr": 2e-4,
    "beta1": 0.5,
    "beta2": 0.999,
    "epochs": 50,   # set to 5 for quick dry run
    "noise_dim": 100,
    "latent_dims_all": [8, 16, 32, 64, 128],
    "latent_dims_compare": [8, 32, 128],
    "kl_anneal_epochs": 10,
    "vae_recon_loss": "bce",  # bce/mse
    "datasets": ["fashionmnist", "emnist_balanced"],
    "eval_samples": 1024,
    "num_workers": 2,
    "max_train_batches_per_epoch": None,  # set e.g. 50 for shorter run
}
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

def seed_everything(seed=42):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)
    if DEVICE == "cuda":
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
def _cuda_startup_ok():
    if not torch.cuda.is_available():
        return False
    try:
        _ = torch.zeros(1, device="cuda")
        torch.cuda.synchronize()
        return True
    except Exception:
        return False

if DEVICE == "cuda" and not _cuda_startup_ok():
    print("CUDA context is in error state. Falling back to CPU. Restart Kaggle session to use GPU again.")
    DEVICE = "cpu"

seed_everything(CFG["seed"])
print("DEVICE:", DEVICE, "| IS_KAGGLE:", IS_KAGGLE)

def _to_device(t):
    return t.to(DEVICE)

def bce_logits_safe(logits, target):
    target = target.float().clamp(0.0, 1.0)
    if torch.isnan(logits).any() or torch.isinf(logits).any():
        raise RuntimeError("NaN/Inf detected in BCE logits")
    return F.binary_cross_entropy_with_logits(logits, target)

def get_num_classes(ds): return 10 if ds=="fashionmnist" else 47
def get_transforms(kind):
    if kind in ["gan","cgan","dcgan"]:
        return transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.5,), (0.5,))])
    return transforms.Compose([transforms.ToTensor()])
def _safe(ds_cls, train, transform, **kwargs):
    try:
        return ds_cls(root=DATA_ROOT, train=train, transform=transform, download=False, **kwargs)
    except Exception:
        if DOWNLOAD_IF_MISSING:
            return ds_cls(root=DOWNLOAD_CACHE, train=train, transform=transform, download=True, **kwargs)
        raise RuntimeError(f"Dataset missing. Mount under {DATA_ROOT} or enable download.")
def get_dataset(ds, train, kind):
    t = get_transforms(kind)
    if ds=="fashionmnist": return _safe(datasets.FashionMNIST, train, t)
    if ds=="emnist_balanced": return _safe(datasets.EMNIST, train, t, split="balanced")
    raise ValueError(ds)
def get_loaders(ds, kind):
    tr, te = get_dataset(ds, True, kind), get_dataset(ds, False, kind)
    return (
        DataLoader(tr, batch_size=CFG["batch_size"], shuffle=True, num_workers=CFG["num_workers"], drop_last=True),
        DataLoader(te, batch_size=CFG["batch_size"], shuffle=False, num_workers=CFG["num_workers"], drop_last=False),
    )

def ensure01(x):
    if x.min() < 0: x = (x + 1)/2
    return x.clamp(0,1)
plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.titleweight": "bold",
    "axes.labelsize": 10,
    "axes.titlesize": 12,
    "legend.frameon": False,
    "grid.alpha": 0.25,
})
def to299(x):
    x = ensure01(x)
    if x.shape[1]==1: x = x.repeat(1,3,1,1)
    return F.interpolate(x, size=(299,299), mode="bilinear", align_corners=False)
def show_grid(x, title, nrow=8, save_name=None):
    g = vutils.make_grid(ensure01(x.detach().cpu())[:nrow*nrow], nrow=nrow)
    fig, ax = plt.subplots(figsize=(7,7))
    ax.imshow(g.permute(1,2,0).squeeze(), cmap="gray")
    ax.axis("off")
    ax.set_title(title, pad=10)
    if save_name:
        fig.savefig(os.path.join(FIG_DIR, save_name), bbox_inches="tight", dpi=180)
    plt.show()
    plt.close(fig)
def plot_hist(hist, title, save_name=None):
    fig, ax = plt.subplots(figsize=(9,4.5))
    for k,v in hist.items(): plt.plot(v, label=k)
    ax.set_title(title)
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Loss")
    ax.grid(True)
    ax.legend()
    if save_name:
        fig.savefig(os.path.join(FIG_DIR, save_name), bbox_inches="tight", dpi=180)
    plt.show()
    plt.close(fig)

class IncepFeat(nn.Module):
    def __init__(self):
        super().__init__()
        try: m = inception_v3(weights="DEFAULT", transform_input=False)
        except Exception: m = inception_v3(weights=None, transform_input=False)
        m.fc = nn.Identity()
        self.m = m.eval()
        for p in self.m.parameters(): p.requires_grad = False
    def forward(self,x): return self.m(x)
def _cov(x):
    x = x - x.mean(axis=0, keepdims=True)
    return (x.T @ x) / max(1, x.shape[0]-1)
def _sqrtm_psd(a):
    w,v = np.linalg.eigh(a); w = np.clip(w,0,None)
    return (v*np.sqrt(w+1e-12))@v.T
def _features(imgs):
    m = IncepFeat().to(DEVICE); out=[]
    with torch.no_grad():
        for i in range(0, len(imgs), 128):
            out.append(m(to299(imgs[i:i+128].to(DEVICE))).cpu())
    return torch.cat(out,0).numpy()
def compute_fid(real,fake):
    r,f = _features(real), _features(fake)
    mr,mf = r.mean(0), f.mean(0)
    cr,cf = _cov(r), _cov(f)
    cm = _sqrtm_psd(cr@cf)
    d = mr-mf
    return float(np.real(d@d + np.trace(cr + cf - 2*cm)))
def compute_is(fake, splits=10):
    try:
        m = inception_v3(weights="DEFAULT").to(DEVICE).eval()
    except Exception:
        m = inception_v3(weights=None).to(DEVICE).eval()
    preds=[]
    with torch.no_grad():
        for i in range(0, len(fake), 128):
            preds.append(F.softmax(m(to299(fake[i:i+128].to(DEVICE))), dim=1).cpu())
    p = torch.cat(preds,0); n = p.size(0); scores=[]
    for k in range(splits):
        part = p[k*n//splits:(k+1)*n//splits]; py = part.mean(0, keepdim=True)
        kl = part*(torch.log(part+1e-8)-torch.log(py+1e-8))
        scores.append(torch.exp(kl.sum(1).mean()).item())
    return float(np.mean(scores)), float(np.std(scores))

def collect_real(loader, n):
    xs=[]
    for x,_ in loader:
        xs.append(x)
        if sum(t.size(0) for t in xs)>=n: break
    return torch.cat(xs,0)[:n]

# ============ Models ============
class VAE(nn.Module):
    def __init__(self, z=32):
        super().__init__()
        self.enc = nn.Sequential(nn.Flatten(), nn.Linear(784,512), nn.ReLU(), nn.Linear(512,256), nn.ReLU())
        self.mu = nn.Linear(256,z); self.logv = nn.Linear(256,z)
        self.dec = nn.Sequential(nn.Linear(z,256), nn.ReLU(), nn.Linear(256,512), nn.ReLU(), nn.Linear(512,784), nn.Sigmoid())
    def encode(self,x):
        h = self.enc(x); return self.mu(h), self.logv(h)
    def rep(self,mu,logv):
        e = torch.randn_like(mu); return mu + e*torch.exp(0.5*logv)
    def decode(self,z): return self.dec(z).view(-1,1,28,28)
    def forward(self,x):
        mu,logv = self.encode(x); z = self.rep(mu,logv); return self.decode(z),mu,logv,z
def vae_loss(recon,x,mu,logv,beta=1.0,kind="bce"):
    if kind == "bce":
        x = x.clamp(0, 1)
        if torch.isnan(recon).any() or torch.isinf(recon).any():
            raise RuntimeError("NaN/Inf detected in VAE reconstruction")
        rec = F.binary_cross_entropy(recon, x, reduction="sum") / x.size(0)
    else:
        rec = F.mse_loss(recon, x, reduction="sum") / x.size(0)
    kl = -0.5*torch.sum(1+logv-mu.pow(2)-logv.exp())/x.size(0)
    return rec + beta*kl, rec, kl

class GFC(nn.Module):
    def __init__(self,nz=100):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(nz,256), nn.LeakyReLU(0.2), nn.Linear(256,512), nn.LeakyReLU(0.2),
                                 nn.Linear(512,1024), nn.LeakyReLU(0.2), nn.Linear(1024,784), nn.Tanh())
    def forward(self,z): return self.net(z).view(-1,1,28,28)
class DFC(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Flatten(), nn.Linear(784,512), nn.LeakyReLU(0.2), nn.Dropout(0.3),
                                 nn.Linear(512,256), nn.LeakyReLU(0.2), nn.Dropout(0.3), nn.Linear(256,1))
    def forward(self,x): return self.net(x)
def gp(d,real,fake):
    a = torch.rand(real.size(0),1,1,1,device=real.device); inter = (a*real+(1-a)*fake).requires_grad_(True)
    o = d(inter)
    g = torch.autograd.grad(o, inter, torch.ones_like(o), create_graph=True, retain_graph=True)[0]
    return ((g.view(g.size(0),-1).norm(2,1)-1)**2).mean()

class CG(nn.Module):
    def __init__(self,nz,nc,ed=32):
        super().__init__(); self.emb=nn.Embedding(nc,ed)
        self.net = nn.Sequential(nn.Linear(nz+ed,256), nn.LeakyReLU(0.2), nn.Linear(256,512), nn.LeakyReLU(0.2),
                                 nn.Linear(512,1024), nn.LeakyReLU(0.2), nn.Linear(1024,784), nn.Tanh())
    def forward(self,z,y): return self.net(torch.cat([z,self.emb(y)],1)).view(-1,1,28,28)
class CD(nn.Module):
    def __init__(self,nc,ed=32):
        super().__init__(); self.emb=nn.Embedding(nc,ed)
        self.net = nn.Sequential(nn.Linear(784+ed,512), nn.LeakyReLU(0.2), nn.Dropout(0.3),
                                 nn.Linear(512,256), nn.LeakyReLU(0.2), nn.Dropout(0.3), nn.Linear(256,1))
    def forward(self,x,y): return self.net(torch.cat([x.view(x.size(0),-1), self.emb(y)],1))

def sn(m, use): return nn.utils.spectral_norm(m) if use else m
class DCG(nn.Module):
    def __init__(self,nz=100,bn=True):
        super().__init__()
        def b(c): return nn.BatchNorm2d(c) if bn else nn.Identity()
        self.net = nn.Sequential(
            nn.ConvTranspose2d(nz,256,7,1,0,bias=False), b(256), nn.ReLU(True),
            nn.ConvTranspose2d(256,128,4,2,1,bias=False), b(128), nn.ReLU(True),
            nn.ConvTranspose2d(128,64,4,2,1,bias=False), b(64), nn.ReLU(True),
            nn.Conv2d(64,1,3,1,1), nn.Tanh()
        )
    def forward(self,z): return self.net(z.view(z.size(0),z.size(1),1,1))
class DCD(nn.Module):
    def __init__(self,use_sn=False):
        super().__init__()
        self.net = nn.Sequential(
            sn(nn.Conv2d(1,64,4,2,1),use_sn), nn.LeakyReLU(0.2,True),
            sn(nn.Conv2d(64,128,4,2,1),use_sn), nn.LeakyReLU(0.2,True),
            sn(nn.Conv2d(128,256,3,2,1),use_sn), nn.LeakyReLU(0.2,True),
            nn.Flatten(), sn(nn.Linear(256*4*4,1),use_sn)
        )
    def forward(self,x): return self.net(x)

class Clf(nn.Module):
    def __init__(self,nc):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1,32,3,1,1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(32,64,3,1,1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Flatten(), nn.Linear(64*7*7,256), nn.ReLU(), nn.Linear(256,nc)
        )
    def forward(self,x): return self.net(x)

# ============ Trainers ============
def _iter_guard(bi):
    return CFG["max_train_batches_per_epoch"] is not None and bi >= CFG["max_train_batches_per_epoch"]

def train_vae(ds,zdim):
    tr,te = get_loaders(ds,"vae"); m=VAE(zdim).to(DEVICE); opt=torch.optim.Adam(m.parameters(),lr=CFG["lr"])
    h={"elbo":[],"recon":[],"kl":[]}; st=time.time()
    for ep in range(CFG["epochs"]):
        el=rl=kl=0.0
        for bi,(x,_) in enumerate(tr):
            if _iter_guard(bi): break
            x=x.to(DEVICE); opt.zero_grad(); recon,mu,logv,_=m(x)
            beta=min(1.0,(ep+1)/max(1,CFG["kl_anneal_epochs"]))
            loss,r,k=vae_loss(recon,x,mu,logv,beta,CFG["vae_recon_loss"]); loss.backward(); opt.step()
            el+=loss.item(); rl+=r.item(); kl+=k.item()
        n=bi+1; h["elbo"].append(el/n); h["recon"].append(rl/n); h["kl"].append(kl/n)
        print(f"[VAE][{ds}][z={zdim}] {ep+1}/{CFG['epochs']} ELBO={h['elbo'][-1]:.4f}")
    return m,h,tr,te,time.time()-st

def train_gan(ds,loss_name="vanilla",nz=None,epochs=None):
    nz = nz or CFG["noise_dim"]; epochs = epochs or CFG["epochs"]
    tr,te = get_loaders(ds,"gan"); G,D=GFC(nz).to(DEVICE),DFC().to(DEVICE)
    og=torch.optim.Adam(G.parameters(),lr=CFG["lr"],betas=(CFG["beta1"],CFG["beta2"]))
    od=torch.optim.Adam(D.parameters(),lr=CFG["lr"],betas=(CFG["beta1"],CFG["beta2"]))
    mse=nn.MSELoss(); h={"D":[],"G":[]}; st=time.time()
    for ep in range(epochs):
        de=ge=0.0
        for bi,(real,_) in enumerate(tr):
            if _iter_guard(bi): break
            real=real.to(DEVICE); bs=real.size(0); z=torch.randn(bs,nz,device=DEVICE); fake=G(z).detach()
            od.zero_grad(); dr,df=D(real),D(fake)
            if loss_name in ["vanilla","non_saturating"]:
                dl=bce_logits_safe(dr,torch.full((bs,1),0.9,device=DEVICE))+bce_logits_safe(df,torch.zeros((bs,1),device=DEVICE))
            elif loss_name=="lsgan":
                dl=mse(torch.sigmoid(dr),torch.ones_like(dr))+mse(torch.sigmoid(df),torch.zeros_like(df))
            elif loss_name in ["wgan","wgan_gp"]:
                dl=-(dr.mean()-df.mean()); 
                if loss_name=="wgan_gp": dl=dl+10.0*gp(D,real,fake)
            else: raise ValueError(loss_name)
            dl.backward(); od.step()
            if loss_name=="wgan":
                for p in D.parameters(): p.data.clamp_(-0.01,0.01)
            og.zero_grad(); z=torch.randn(bs,nz,device=DEVICE); dg=D(G(z))
            if loss_name=="vanilla": gl=bce_logits_safe(dg,torch.ones((bs,1),device=DEVICE))
            elif loss_name=="non_saturating": gl=-torch.log(torch.sigmoid(dg)+1e-8).mean()
            elif loss_name=="lsgan": gl=mse(torch.sigmoid(dg),torch.ones_like(dg))
            else: gl=-dg.mean()
            gl.backward(); og.step()
            de+=dl.item(); ge+=gl.item()
        n=bi+1; h["D"].append(de/n); h["G"].append(ge/n)
        print(f"[GAN:{loss_name}][{ds}] {ep+1}/{epochs} D={h['D'][-1]:.4f} G={h['G'][-1]:.4f}")
    return G,D,h,tr,te,time.time()-st

def train_cgan(ds):
    tr,te = get_loaders(ds,"cgan"); nc=get_num_classes(ds); G,D=CG(CFG["noise_dim"],nc).to(DEVICE),CD(nc).to(DEVICE)
    og=torch.optim.Adam(G.parameters(),lr=CFG["lr"],betas=(CFG["beta1"],CFG["beta2"]))
    od=torch.optim.Adam(D.parameters(),lr=CFG["lr"],betas=(CFG["beta1"],CFG["beta2"]))
    h={"D":[],"G":[]}; st=time.time()
    for ep in range(CFG["epochs"]):
        de=ge=0.0
        for bi,(real,y) in enumerate(tr):
            if _iter_guard(bi): break
            real,y=real.to(DEVICE),y.to(DEVICE); bs=real.size(0)
            z=torch.randn(bs,CFG["noise_dim"],device=DEVICE); fake=G(z,y).detach()
            od.zero_grad(); dl=bce_logits_safe(D(real,y),torch.full((bs,1),0.9,device=DEVICE))+bce_logits_safe(D(fake,y),torch.zeros((bs,1),device=DEVICE)); dl.backward(); od.step()
            og.zero_grad(); z=torch.randn(bs,CFG["noise_dim"],device=DEVICE); gl=bce_logits_safe(D(G(z,y),y),torch.ones((bs,1),device=DEVICE)); gl.backward(); og.step()
            de+=dl.item(); ge+=gl.item()
        n=bi+1; h["D"].append(de/n); h["G"].append(ge/n); print(f"[CGAN][{ds}] {ep+1}/{CFG['epochs']} D={h['D'][-1]:.4f} G={h['G'][-1]:.4f}")
    return G,D,h,tr,te,time.time()-st

def train_dcgan(ds,bn=True,use_sn=False,epochs=None):
    epochs = epochs or CFG["epochs"]
    tr,te=get_loaders(ds,"dcgan"); G,D=DCG(CFG["noise_dim"],bn).to(DEVICE),DCD(use_sn).to(DEVICE)
    og=torch.optim.Adam(G.parameters(),lr=CFG["lr"],betas=(CFG["beta1"],CFG["beta2"]))
    od=torch.optim.Adam(D.parameters(),lr=CFG["lr"],betas=(CFG["beta1"],CFG["beta2"]))
    h={"D":[],"G":[]}; st=time.time()
    for ep in range(epochs):
        de=ge=0.0
        for bi,(real,_) in enumerate(tr):
            if _iter_guard(bi): break
            real=real.to(DEVICE); bs=real.size(0); z=torch.randn(bs,CFG["noise_dim"],device=DEVICE); fake=G(z).detach()
            od.zero_grad(); dl=bce_logits_safe(D(real),torch.full((bs,1),0.9,device=DEVICE))+bce_logits_safe(D(fake),torch.zeros((bs,1),device=DEVICE)); dl.backward(); od.step()
            og.zero_grad(); z=torch.randn(bs,CFG["noise_dim"],device=DEVICE); gl=bce_logits_safe(D(G(z)),torch.ones((bs,1),device=DEVICE)); gl.backward(); og.step()
            de+=dl.item(); ge+=gl.item()
        n=bi+1; h["D"].append(de/n); h["G"].append(ge/n); print(f"[DCGAN][{ds}] {ep+1}/{epochs} D={h['D'][-1]:.4f} G={h['G'][-1]:.4f} bn={bn} sn={use_sn}")
    return G,D,h,tr,te,time.time()-st

def train_clf(ds,fraction=1.0,epochs=3,extra=None):
    tr,te=get_loaders(ds,"vae"); base=tr.dataset; n=int(len(base)*fraction); idx=np.random.permutation(len(base))[:n]
    ds_sub=Subset(base,idx)

    class _TensorLabelSubset(Dataset):
        def __init__(self, subset):
            self.subset = subset

        def __len__(self):
            return len(self.subset)

        def __getitem__(self, index):
            image, label = self.subset[index]
            if not torch.is_tensor(label):
                label = torch.tensor(label, dtype=torch.long)
            return image, label.long()

    if extra is not None:
        xs,ys=extra

        class _SyntheticLabeledDataset(Dataset):
            def __init__(self, images, labels):
                self.images = images
                self.labels = labels

            def __len__(self):
                return self.images.size(0)

            def __getitem__(self, index):
                return self.images[index], self.labels[index].long()

        ds_sub=torch.utils.data.ConcatDataset([_TensorLabelSubset(ds_sub), _SyntheticLabeledDataset(xs.cpu(), ys.cpu())])
    else:
        ds_sub=_TensorLabelSubset(ds_sub)
    dl=DataLoader(ds_sub,batch_size=CFG["batch_size"],shuffle=True,num_workers=CFG["num_workers"])
    m=Clf(get_num_classes(ds)).to(DEVICE); opt=torch.optim.Adam(m.parameters(),lr=1e-3); ce=nn.CrossEntropyLoss()
    for _ in range(epochs):
        m.train()
        for x,y in dl:
            x,y=x.to(DEVICE),y.to(DEVICE); opt.zero_grad(); loss=ce(m(x),y); loss.backward(); opt.step()
    m.eval(); c=t=0
    with torch.no_grad():
        for x,y in te:
            x,y=x.to(DEVICE),y.to(DEVICE); p=m(x).argmax(1); c+=(p==y).sum().item(); t+=y.numel()
    return m, c/max(1,t)

def cgan_cls_acc(G, ds, n_per_class=64):
    nc=get_num_classes(ds); gen_nc=int(G.emb.num_embeddings); use_nc=min(nc,gen_nc)  # avoids CUDA label assert
    clf,_=train_clf(ds,1.0,2)
    xs,ys=[],[]
    with torch.no_grad():
        for c in range(use_nc):
            z=torch.randn(n_per_class,CFG["noise_dim"],device=DEVICE); y=torch.full((n_per_class,),c,dtype=torch.long,device=DEVICE)
            xs.append(ensure01(G(z,y)).cpu()); ys.append(y.cpu())
    xs,ys=torch.cat(xs,0),torch.cat(ys,0)
    with torch.no_grad():
        p=clf(xs.to(DEVICE)).argmax(1).cpu()
    return float((p==ys).float().mean().item())

def sample_from(model_name, model, ds, n):
    if model_name=="VAE":
        with torch.no_grad():
            z=torch.randn(n,model.mu.out_features,device=DEVICE); x=model.decode(z)
        y=torch.randint(0,get_num_classes(ds),(n,)); return x.cpu(),y
    if model_name=="GAN":
        with torch.no_grad(): x=model(torch.randn(n,CFG["noise_dim"],device=DEVICE))
        y=torch.randint(0,get_num_classes(ds),(n,)); return ensure01(x).cpu(),y
    if model_name=="CGAN":
        y=torch.randint(0,get_num_classes(ds),(n,),device=DEVICE)
        with torch.no_grad(): x=model(torch.randn(n,CFG["noise_dim"],device=DEVICE),y)
        return ensure01(x).cpu(),y.cpu()
    if model_name=="DCGAN":
        with torch.no_grad(): x=model(torch.randn(n,CFG["noise_dim"],device=DEVICE))
        y=torch.randint(0,get_num_classes(ds),(n,)); return ensure01(x).cpu(),y
    raise ValueError(model_name)

def mode_score(x): return float(torch.pdist(ensure01(x).view(x.size(0),-1),p=2).mean().item())
def checker_score(x):
    x=ensure01(x); dx=torch.abs(x[:,:,:,1:]-x[:,:,:,:-1]).mean(); dy=torch.abs(x[:,:,1:,:]-x[:,:,:-1,:]).mean()
    return float(((dx+dy)/2).item())

def smoke():
    x=torch.rand(8,1,28,28,device=DEVICE); y=torch.randint(0,10,(8,),device=DEVICE)
    v=VAE(16).to(DEVICE); r,mu,lv,_=v(x); _=vae_loss(r,x,mu,lv)
    g,d=GFC(CFG["noise_dim"]).to(DEVICE),DFC().to(DEVICE); _=d(g(torch.randn(8,CFG["noise_dim"],device=DEVICE)))
    cg,cd=CG(CFG["noise_dim"],10).to(DEVICE),CD(10).to(DEVICE); _=cd(cg(torch.randn(8,CFG["noise_dim"],device=DEVICE),y),y)
    dg,dd=DCG(CFG["noise_dim"]).to(DEVICE),DCD().to(DEVICE); _=dd(dg(torch.randn(8,CFG["noise_dim"],device=DEVICE)))
    print("Smoke tests passed.")

def preflight_device():
    global DEVICE
    if DEVICE != "cuda":
        return
    try:
        smoke()
    except Exception as e:
        print("CUDA preflight failed. Falling back to CPU. Error:", e)
        DEVICE = "cpu"
        smoke()

def run_all():
    preflight_device()
    results={"vae":{}, "gan":{}, "cgan":{}, "dcgan":{}, "loss_variants":{}, "augmentation":{}}
    summary=[]
    for ds in CFG["datasets"]:
        print("\n=== DATASET:", ds, "===\n")
        # VAE suite
        for z in CFG["latent_dims_all"]:
            m,h,tr,te,tsec=train_vae(ds,z); plot_hist({"ELBO":h["elbo"],"Recon":h["recon"],"KL":h["kl"]},f"VAE ({ds}, z={z})",f"vae_loss_{ds}_z{z}.png")
            xb,yb=next(iter(te)); xb=xb.to(DEVICE)
            with torch.no_grad(): recon,mu,lv,_=m(xb)
            fake=sample_from("VAE",m,ds,CFG["eval_samples"])[0]; fid=compute_fid(collect_real(te,CFG["eval_samples"]),fake,); ssim=float(structural_similarity_index_measure(ensure01(recon.cpu()),ensure01(xb.cpu())).item())
            summary.append({"dataset":ds,"model":"VAE","latent_dim":z,"recon_loss":float(h["recon"][-1]),"kl":float(h["kl"][-1]),"fid":fid,"ssim":ssim,"train_time_sec":tsec})
            results["vae"][(ds,z)]={"model":m,"hist":h}
            if z in CFG["latent_dims_compare"]: show_grid(fake,f"VAE random sampling ({ds}, z={z})",save_name=f"vae_samples_{ds}_z{z}.png")
            # interpolation
            with torch.no_grad():
                mu1,_=m.encode(xb[:1]); mu2,_=m.encode(xb[1:2]); a=torch.linspace(0,1,10,device=DEVICE).unsqueeze(1); inter=m.decode((1-a)*mu1+a*mu2)
            show_grid(inter,f"VAE interpolation ({ds}, z={z})",nrow=10,save_name=f"vae_interp_{ds}_z{z}.png")
            # conditional latent control
            class_mu={c:[] for c in range(get_num_classes(ds))}
            with torch.no_grad():
                for x,y in te:
                    mu_,_=m.encode(x.to(DEVICE))
                    for i in range(x.size(0)): class_mu[int(y[i])].append(mu_[i].cpu())
            keys=[k for k,v in class_mu.items() if len(v)>0]
            if len(keys)>=2:
                zc=torch.stack([torch.stack(class_mu[keys[0]]).mean(0), torch.stack(class_mu[keys[-1]]).mean(0)],0).to(DEVICE)
                with torch.no_grad(): xc=m.decode(zc)
                show_grid(xc,f"VAE conditional control ({ds}, z={z})",nrow=2,save_name=f"vae_control_{ds}_z{z}.png")
            # tsne/umap
            try:
                z2=TSNE(n_components=2, init="random", learning_rate="auto").fit_transform(mu.detach().cpu().numpy())
                plt.figure(figsize=(5,4)); plt.scatter(z2[:,0],z2[:,1],c=yb[:len(z2)],s=8,cmap="tab20"); plt.title(f"t-SNE ({ds}, z={z})")
                plt.savefig(os.path.join(FIG_DIR,f"tsne_{ds}_z{z}.png"),bbox_inches="tight",dpi=180); plt.show()
            except Exception as e: print("t-SNE skipped", e)
            try:
                z3=umap.UMAP(n_components=2).fit_transform(mu.detach().cpu().numpy())
                plt.figure(figsize=(5,4)); plt.scatter(z3[:,0],z3[:,1],c=yb[:len(z3)],s=8,cmap="tab20"); plt.title(f"UMAP ({ds}, z={z})")
                plt.savefig(os.path.join(FIG_DIR,f"umap_{ds}_z{z}.png"),bbox_inches="tight",dpi=180); plt.show()
            except Exception as e: print("UMAP skipped", e)

        # Vanilla GAN
        G,D,h,tr,te,tsec=train_gan(ds,"vanilla",CFG["noise_dim"],CFG["epochs"]); plot_hist(h,f"Vanilla GAN ({ds})",f"gan_loss_{ds}.png")
        with torch.no_grad(): fake=G(torch.randn(CFG["eval_samples"],CFG["noise_dim"],device=DEVICE)).cpu()
        show_grid(fake,f"Vanilla GAN samples ({ds})",save_name=f"gan_samples_{ds}.png")
        summary.append({"dataset":ds,"model":"GAN","fid":compute_fid(collect_real(te,CFG["eval_samples"]),fake),"mode_diversity":mode_score(fake),"checkerboard":checker_score(fake),"train_time_sec":tsec})
        results["gan"][ds]={"G":G,"D":D,"hist":h}
        noise_rows=[]
        for nz in [32,64,100,128]:
            G2,_,_,_,te2,_=train_gan(ds,"vanilla",nz,max(5,CFG["epochs"]//5))
            with torch.no_grad(): f2=G2(torch.randn(CFG["eval_samples"],nz,device=DEVICE)).cpu()
            noise_rows.append({"dataset":ds,"noise_dim":nz,"fid":compute_fid(collect_real(te2,CFG["eval_samples"]),f2)})
        pd.DataFrame(noise_rows).to_csv(os.path.join(TAB_DIR,f"noise_dim_study_{ds}.csv"),index=False)
        # gradient penalty experiment
        _,_,hgp,_,_,_=train_gan(ds,"wgan_gp",CFG["noise_dim"],max(5,CFG["epochs"]//5)); plot_hist(hgp,f"Gradient penalty experiment ({ds})",f"gan_gp_{ds}.png")

        # CGAN
        cG,cD,ch,tr,te,tsec=train_cgan(ds); plot_hist(ch,f"CGAN ({ds})",f"cgan_loss_{ds}.png")
        nc=get_num_classes(ds)
        for c in range(min(10,nc)):
            with torch.no_grad(): x=cG(torch.randn(64,CFG["noise_dim"],device=DEVICE), torch.full((64,),c,dtype=torch.long,device=DEVICE))
            show_grid(x,f"CGAN class {c} ({ds})",save_name=f"cgan_{ds}_class_{c}.png")
        with torch.no_grad():
            z=torch.randn(16,CFG["noise_dim"],device=DEVICE); x1=cG(z,torch.zeros(16,dtype=torch.long,device=DEVICE)); x2=cG(z,torch.ones(16,dtype=torch.long,device=DEVICE)); a=torch.linspace(0,1,16,device=DEVICE).view(16,1,1,1); inter=(1-a)*x1+a*x2
        show_grid(inter,f"CGAN label interpolation ({ds})",nrow=8,save_name=f"cgan_interp_{ds}.png")
        with torch.no_grad():
            n=CFG["eval_samples"]; z=torch.randn(n,CFG["noise_dim"],device=DEVICE); y=torch.randint(0,nc,(n,),device=DEVICE); yw=(y+1)%nc
            ft=cG(z,y).cpu(); fw=cG(z,yw).cpu()
        real=collect_real(te,CFG["eval_samples"])
        class_fids=[]
        for c in range(nc):
            with torch.no_grad(): fc=cG(torch.randn(256,CFG["noise_dim"],device=DEVICE), torch.full((256,),c,dtype=torch.long,device=DEVICE)).cpu()
            class_fids.append(compute_fid(real[:256],fc))
        is_m,is_s=compute_is(ft)
        summary.append({"dataset":ds,"model":"CGAN","fid":compute_fid(real,ft),"fid_mismatch":compute_fid(real,fw),"intra_fid":float(np.mean(class_fids)),"classifier_acc":cgan_cls_acc(cG,ds),"is_mean":is_m,"is_std":is_s,"train_time_sec":tsec})
        results["cgan"][ds]={"G":cG,"D":cD,"hist":ch}

        # DCGAN
        dG,dD,dh,tr,te,tsec=train_dcgan(ds,True,False,CFG["epochs"]); plot_hist(dh,f"DCGAN ({ds})",f"dcgan_loss_{ds}.png")
        with torch.no_grad(): df=dG(torch.randn(CFG["eval_samples"],CFG["noise_dim"],device=DEVICE)).cpu()
        show_grid(df,f"DCGAN samples ({ds})",save_name=f"dcgan_samples_{ds}.png")
        dG2,_,_,_,te2,_=train_dcgan(ds,False,False,max(5,CFG["epochs"]//5)); 
        with torch.no_grad(): df2=dG2(torch.randn(CFG["eval_samples"],CFG["noise_dim"],device=DEVICE)).cpu()
        dG3,_,_,_,te3,_=train_dcgan(ds,True,True,max(5,CFG["epochs"]//5)); 
        with torch.no_grad(): df3=dG3(torch.randn(CFG["eval_samples"],CFG["noise_dim"],device=DEVICE)).cpu()
        with torch.no_grad():
            z1,z2,z3=[torch.randn(16,CFG["noise_dim"],device=DEVICE) for _ in range(3)]; la=dG(z1-z2+z3)
        show_grid(la,f"DCGAN latent arithmetic ({ds})",save_name=f"dcgan_latent_arith_{ds}.png")
        summary.append({"dataset":ds,"model":"DCGAN","fid":compute_fid(collect_real(te,CFG["eval_samples"]),df),"fid_no_bn":compute_fid(collect_real(te2,CFG["eval_samples"]),df2),"fid_sn":compute_fid(collect_real(te3,CFG["eval_samples"]),df3),"fid_progressive_proxy":compute_fid(collect_real(te,CFG["eval_samples"]),df),"train_time_sec":tsec})
        results["dcgan"][ds]={"G":dG,"D":dD,"hist":dh}

        # Loss variants
        lv=[]
        for ln in ["vanilla","non_saturating","wgan","wgan_gp","lsgan"]:
            g,_,hh,_,te_,tt=train_gan(ds,ln,CFG["noise_dim"],max(5,CFG["epochs"]//3)); plot_hist(hh,f"Loss variant {ln} ({ds})",f"loss_variant_{ln}_{ds}.png")
            with torch.no_grad(): ff=g(torch.randn(CFG["eval_samples"],CFG["noise_dim"],device=DEVICE)).cpu()
            lv.append({"dataset":ds,"loss_variant":ln,"fid":compute_fid(collect_real(te_,CFG["eval_samples"]),ff),"train_time_sec":tt,"stability_proxy":float(np.std(hh["D"])+np.std(hh["G"]))})
        pd.DataFrame(lv).to_csv(os.path.join(TAB_DIR,f"loss_variants_{ds}.csv"),index=False); results["loss_variants"][ds]=lv

        # Augmentation study
        _,base=train_clf(ds,0.10,5); aug=[{"dataset":ds,"generator":"None","synthetic_ratio":0.0,"acc":base}]
        real_count=int(len(get_dataset(ds,True,"vae"))*0.10)
        pool=[]
        if (ds,32) in results["vae"]: pool.append(("VAE",results["vae"][(ds,32)]["model"]))
        pool.append(("GAN",results["gan"][ds]["G"])); pool.append(("CGAN",results["cgan"][ds]["G"])); pool.append(("DCGAN",results["dcgan"][ds]["G"]))
        for n,g in pool:
            for ratio in [0.25,0.50,0.75,1.00]:
                xs,ys=sample_from(n,g,ds,int(real_count*ratio))
                _,acc=train_clf(ds,0.10,5,extra=(xs,ys))
                aug.append({"dataset":ds,"generator":n,"synthetic_ratio":ratio,"acc":acc})
                print(f"[AUG][{ds}] {n} ratio={ratio:.2f} acc={acc:.4f}")
        adf=pd.DataFrame(aug); adf.to_csv(os.path.join(TAB_DIR,f"augmentation_{ds}.csv"),index=False); results["augmentation"][ds]=adf

        # cross-dataset transfer (safe labels to avoid CUDA assert)
        other = "emnist_balanced" if ds=="fashionmnist" else "fashionmnist"
        try:
            acc=cgan_cls_acc(results["cgan"][ds]["G"], other, 64)
            print(f"Cross-dataset transfer classifier accuracy (gen on {ds}, eval on {other}): {acc:.4f}")
        except Exception as e:
            print("Cross-dataset transfer skipped:", e)

        comp = pd.DataFrame(summary)
        comp.to_csv(os.path.join(TAB_DIR,"comparative_results.csv"),index=False)
        print("\nSaved all deliverables to:", OUT_DIR)
        print("Figures in:", FIG_DIR)
        print("Tables in:", TAB_DIR)
    print(comp.head())
    return comp, results

comp, results = run_all()
```
